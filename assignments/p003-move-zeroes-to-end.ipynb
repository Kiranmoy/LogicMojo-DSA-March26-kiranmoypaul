# Problem

Given an integer array nums, move all 0's to the end of it while maintaining the relative order of the non-zero elements.

> Note that you must do this in-place without making a copy of the array.

**Input Format**: 
* First line is the length of the array
* Next line is the elements of the array

**Constraints**
* `1 <= arr.length <= 104`
* `0 <= arr[i] <= 231 - 1`

**Output Format**: Return the array as per the requirement

**Sample Input 0**

```
6
1 2 0 0 3 4
```

**Sample Output 0**

```
1 2 3 4 0 0
```

# Pattern / Category

* **Two Pointers** (a.k.a. **Stable Compaction**)
* This is a specialized **in-place partitioning** where the order of the “kept” elements (non-zeros) must be preserved.

# Intuition

You want to:
* **Collect non-zeros to the front** (preserving order),
* **Push zeros to the back** (implicitly or explicitly).

Think of it as **writing** non-zero items to the earliest available index (write) while you **scan** (read) through the array.

# Technique: Overwrite + Fill Zeros

* Maintain two pointers, `p1` & `p2`
    * `p1` to track the index where a **non-zero** value will be overwritten.
    * `p2` to scan the element left to right.
* Loop until `p2` < length of the array
    * if `p2` index has a value > 0 then
        * Overwrite `p1` index with value at `p2` index
        * `p1++`
        * `p2++`
    * else if `p1` index has value 0 then
        *  `p2++` 
* After one pass, all **non-zero** values will be shifted left.
* Now fill the rest of the remaining index (until `p1` equals the end index) with zeros.

# Solution

````python
def move_func(n, arr):
    p1 = 0
    p2 = 0

    while p2 < n:
        if arr[p2] > 0:
            arr[p1] = arr[p2]
            p1 = p1 + 1
            p2 = p2 + 1
        elif arr[p2] == 0:
            p2 = p2 + 1

    while p1 < n:
        arr[p1] = 0
        p1 = p1 + 1


arr = [1, 2, 0, 0, 3, 4]
n = len(arr)
move_func(n, arr)
print(arr)
````
