# Problem

Given an array of size N containing only 0s, 1s, and 2s, sort the array in ascending order.

**Input Format**
* Integer N.
* Integer Array of size N.

**Constraints**
* `1 <= N <= 10^6`
* `0 <= A[i] <= 2`

**Output Format**: Integer Array of size N.

**Sample Input 0**

```
5
0 1 2 1 2
```

**Sample Output 0**

```
0 1 1 2 2
```

**Explanation 0**: 0s 1s, and 2s are segregated into ascending order.

# Pattern / Category

* This is a classic **Three-Way Partitioning** problem → also known as the **Dutch National Flag (DNF)** Algorithm.
* It uses the **Two-pointer technique**, but extended into **three regions**.

# Intuition

Instead of counting or sorting normally, we divide the array into **3 logical sections**:





We maintain three pointers:

▶ **low (p1)**
* Marks the boundary where 0s should go
* Everything left of low is 0

▶ **mid (p2)**
* Current pointer scanning the array
* Decides where each element belongs

▶ **high (p3)**
* Marks the boundary where 2s should go
* Everything right of high is 2

# Technique

We scan once `(O(N))` using rules:

**Case 1: `arr[mid] == 0`**
* Swap `arr[mid] ↔ arr[low]`
* Move both `low++` and `mid++`
* Because `0` belongs to the left

**Case 2: `arr[mid] == 1`**
* This element is already in the correct place
* Simply do: `mid++`

**Case 3: `arr[mid] == 2`**
* Swap `arr[mid] ↔ arr[high]`
* Move only `high--`
* Do NOT move `mid` because the swapped element is unknown ***(mid must examine the new element)***

# Solution

````python
def sort012(arr):
    low, mid, high = 0, 0, len(arr) - 1
    
    while mid <= high:
        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1
        
        elif arr[mid] == 1:
            mid += 1
        
        else:  # arr[mid] == 2
            arr[high], arr[mid] = arr[mid], arr[high]
            high -= 1
    
    return arr

arr = [0, 1, 2, 1, 2]
sort012(arr)
print(arr)
````
