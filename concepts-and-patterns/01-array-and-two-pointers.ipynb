# Arrays

* An array is a collection of items (numbers, characters, etc.) that are stored at contiguous memory locations.
* Array always starts from index 0.
* Best sorting techniques are: **Heap sort** & **Merge sort** having worst time complexity as **`O(n logn)`**.

# Two Pointer Approach

1. In this technique, you'll have **two pointers** to store the index location of the first and last element.
2. Depending on the logic you choose, either you have to
   * **increment the first pointer**,
   * or **decrement the last pointer**,
   * or do both.
3. Continue step no. 2 until the `first point >= last pointer` index location.

> **NOTE**: To apply the two-pointer technique, the element must have some correlation from **left to right** & **right to left**. 
>
> For example, take a sorted array of numbers in ascending order:
> * **correlation left to right**: increasing order
> * **correlation right to left**: decreasing order

This correlation is important, as it will help to decide whether we have to
* **increment the first pointer**,
* or **decrement the last pointer**,
* or do both.

Other use cases of two-pointers are:
* Reversing an array
* Checking palindrome




> The main significance of **TWO POINTER** approach is that we are able to track two elements simultaneously.

# Problem 1 - Pair Sum exist

Given a sorted array A (sorted in ascending order), having N integers, find if there exists any pair of elements (`A[i], A[j]`) such that their sum is equal to X.

**For Example**: `A[] = [10, 20, 35, 50, 75, 80]` and the value of `X = 110`.

# Solution 1

````python
def isPairSumExist(arr, targetSum):
    # represents first pointer
    p1 = 0
    # represents second pointer
    p2 = len(arr) - 1

    while p1 < p2:

        # If we find a pair
        if arr[p1] + arr[p2] == targetSum:
            return True

        # If sum of elements at current
        # pointers is less, we move towards
        # higher values by doing p1++
        elif arr[p1] + arr[p2] < targetSum:
            p1 = p1 + 1

        # If sum of elements at current
        # pointers is more, we move towards
        # lower values by doing p2--
        else:
            p2 = p2 - 1

    return False


arr = [10, 20, 35, 50, 75, 80]
targetSum = 110
print(isPairSumExist(arr, targetSum))
````

* **Time Complexity**: `O(N)`
* **Space Complexity**: `O(1)`

# Problem 2 - Find Triplets 

Given an array of distinct elements. The task is to find triplets in the array whose sum is zero.

**Example 1:** 

```
Input: arr = [0, -1, 2, -3, 1]
Output: (0 -1 1), (2 -3 1)
```

**Explanation**: The triplets with zero sum are `0 + -1 + 1 = 0 and 2 + -3 + 1 = 0 `

**Example 2:** 

```
Input: arr = [1, -2, 1, 0, 5]
Output: 1 -2  1
```

**Explanation**: The triplets with a zero sum is `1 + -2 + 1 = 0`   


# Solution 2

````python
def findTriplets(arr, targetSum):
    arr.sort()

    isTripletFound = False

    for i in range(len(arr)):
        fixed_element = arr[i]         # Fixed element
        p1 = i + 1                     # First pointer (moving pointer)
        p2 = len(arr) - 1               # Second pointer (moving pointer)

        while p1 < p2:
            # print elements if it's sum is equal to targetSum
            if (fixed_element + arr[p1] + arr[p2]) == targetSum:
                print("Triplet Indexes:",i, p1, p2)
                print("Triplet Elements:",fixed_element, arr[p1], arr[p2])
                p1 = p1 + 1
                p2 = p2 - 1
                isTripletFound = True

            # If sum of three elements is less than targetSum then increment in left
            elif (fixed_element + arr[p1] + arr[p2]) < targetSum:
                p1 = p1 + 1

            # if sum is greater than targetSum then decrement in right side
            else:
                p2 = p2 - 1

        if not isTripletFound:
            print("No Such Triplet")

arr = [0, -1, 2, -3, 1]
targetSum = 0
findTriplets(arr, targetSum)
````

* **Time Complexity**: `O(N^2)`
* **Space Complexity**: `O(1)`
