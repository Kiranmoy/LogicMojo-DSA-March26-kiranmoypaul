# Problem - Largest Sum Subarray

Given an array `arr[]` of size N. The task is to find the sum of the contiguous subarray within an `arr[]` with the largest sum.

```
A = [-2, -3, 4, -1, -2, 1, 5, -3]`
```
* Maximum contiguous sum is 7
* Starting index 2
* Ending index 6


# Pattern: Kadane’s Algorithm → Running Sum + Reset Technique

At each element:
* Either **extend** the previous subarray
* Or **start a new subarray** from the current element
Whichever gives a larger sum.

This gives a linear-time solution: `O(n)`.

# Intuition 

**Think of it like walking through the array:**
* You carry a running sum (`current_sum`).
* If this running sum ever becomes **negative**,
* it’s harmful — it will only reduce future totals.
* So you **drop it immediately** and start fresh.

**Why drop the negative sum?**
* If your current bag contains a negative total, adding it to future numbers only hurts you.
* So the moment `sum < 0` → throw that bag away.

This is why Kadane’s works beautifully.

> ***A negative prefix can never contribute to a maximum subarray. So reset whenever the running sum goes negative.***

# Solution

````python
def max_subarray_sum(arr):
    max_sum = arr[0]
    current_sum = arr[0]
    start = 0
    end = 0
    potential_starting_index = 0

    for i in range(1, len(arr)):

        current_sum = current_sum + arr[i]

        if current_sum > max_sum:
            max_sum = current_sum
            start = potential_starting_index  # updating start when a new max is found
            end = i

        if current_sum < 0:
            current_sum = 0
            potential_starting_index = i + 1 # Reset new start index

    print("Maximum contiguous sum is", max_sum)
    print("Starting index", start)
    print("Ending index", end)


A = [-2, -3, 4, -1, -2, 1, 5, -3]
max_subarray_sum(A)
````
