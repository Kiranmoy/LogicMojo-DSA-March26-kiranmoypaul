# Index manipulation Technique

It will be used in the problems where you need to **RESTRUCTURE** or **REORDER** the input array elements without using additional space.

# Problem 1: Return Unique Elements length (Remove Duplicates)

Given a sorted array, remove all duplicates and return the length of the modified array solution.

**Examples**: 

```
Input: nums = [1, 1, 2, 2, 2, 3, 4, 4]
Output: 4
```

# Solution 1

````python
def getUniqueElementsLength(arr):
    uniqueElements = 0

    # Start traversing from 2nd element
    for i in range(0, len(arr)):
        if (i == 0) or (arr[i] > arr[i - 1]):
            arr[uniqueElements] = arr[i]  # modifying the existing array
            uniqueElements = uniqueElements + 1

    print("Modified arr without duplicates: ", arr[:uniqueElements]) # print only the modfied part of the array
    return uniqueElements

arr = [1, 1, 2, 2, 2, 3, 4, 4]
print(getUniqueElementsLength(arr))
````

* **Time Complexity**: `O(N)`
* **Space Complexity**: `O(1)`

# Problem 2: Move Zero to the end

Given an array of integers, move all the zeros to the end of the array while maintaining the relative order of the other elements. 

**Input**: `arr = [10,23,0,53,0,21,0]`

**Output**: `arr = [10,23,53,21,0,0,0]`


# Solution 2

````python
def moveZeroesToEnd(arr):
    i = 0
    for item in arr:
        if item != 0:
            arr[i] = item
            i = i + 1

    while i < (len(arr) - 1):
        arr[i] = 0
        i = i + 1

arr = [10,23,0,53,0,21,0]
moveZeroesToEnd(arr) # in-place modification
print(arr)
````

* **Time Complexity**: `O(N)`
* **Space Complexity**: `O(1)`
