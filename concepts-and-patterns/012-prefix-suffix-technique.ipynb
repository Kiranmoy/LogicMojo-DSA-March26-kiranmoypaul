# Problem - Product of Array Except Self

Given an integer array nums, return an array answer such that `answer[i]` is equal to the product of all the elements of nums except `nums[i]`.

The product of any prefix or suffix of nums is guaranteed to fit in a **32-bit** integer.

You must write an algorithm that runs in `O(n)` time and without using the division operation.

**Input**: `nums = [1,2,3,4]`

**Output**: `[24,12,8,6]`


# 🧠 Problem Pattern

This falls under the **Prefix–Suffix** (a.k.a. **Left–Right accumulators**) pattern:
* Build an array of **prefix products** so that `prefix[i]` = product of all elements **before i**.
* Build an array of **suffix products** so that `suffix[i]` = product of all elements **after i**.
* Then, the answer at each index is simply: **`answer[i] = prefix[i] * suffix[i]`** because:
    * `prefix[i]` gives the product of all elements **before i**
    * `suffix[i]` gives the product of all elements **after i**
    * product of `prefix[i] * suffix[i]` give the product of all elements **except element at index i** (`arr[i]`)

> This pattern generalizes to many problems where each position’s answer depends on **all other positions except itself**, and multiplication is associative.

# 💡 Intuition

At index i, we need the **product of all values except** `nums[i]`. 

Think of splitting the array at i:
* Everything to the **left** → multiply them all once.
* Everything to the **right** → multiply them all once.
* Then combine them: **Left product** × **Right product**.

We avoid division to handle cases with zeros and to meet the constraint.

# Technique (Two-Array Version)

1. **Prefix pass** (left → right):

```
Initialize prefix[0] = 1 (no elements to the left of index 0).
prefix[i] = nums[0] * nums[1] * ... * nums[i-1]
```

2. **Suffix pass** (right → left):

```
suffix[i] = nums[i+1] * nums[i+2] * ... * nums[n-1]
Initialize suffix[n-1] = 1 (no elements to the right of last index).
```

3. **Combine**: `result[i] = prefix[i] * suffix[i]`

---

**Time Complexity**: `O(n)`

**Space Complexity**: `O(n)` for prefix + `O(n)` for suffix + `O(n)` for result (can be optimized).

# ♻️ Space-Optimized Technique (O(1) extra space)

We can avoid the extra prefix and suffix arrays (besides the output):

**First pass** (left → right): store prefix products directly into result:

```
result[i] = product of all elements to the left of i
```

**Second pass** (right → left): maintain a running suffix product and multiply into result[i]:

```
result[i] *= current_suffix; 
update current_suffix *= nums[i]
```

# ⚠️ Edge Cases & Zero Handling

* **No zeros**: Use prefix/suffix as above.
* **Exactly one zero**: All positions except the zero’s index will be 0; at the zero’s index, the result is the product of all non-zero elements.
* **Two or more zeros**: Every result is 0.

The prefix–suffix method naturally handles zeros correctly without any special division logic.

# ✅ When to Use This Pattern

Use the **prefix–suffix accumulators** when:
* The result at index `i` depends on **all other indices (except `i`)** in an **associative operation** (e.g., product or sum).
* Division is disallowed or unsafe (e.g., zeros present).
* You need `O(n)` time and `O(1)` extra space (excluding the output).

**Common relatives of this pattern:**
* “Sum of array except self” (trivial with totals, but can also use prefix–suffix).
* “Trapping Rain Water” (max-left and max-right arrays).
* “Best Sightseeing Pair” (max-left transforms).
* “Max product of three numbers” (track top-3 and bottom-2, conceptually similar left/right thinking).

# Solution

````python
def product_except_self_prefix_suffix(nums):
    """
    Return an array where result[i] is the product of all elements of nums except nums[i], 
    using explicit prefix and suffix arrays (no division).

    Time: O(n)
    Space: O(n) extra (for prefix and suffix)
    """
    n = len(nums)
    prefix = [1] * n   # prefix[i] = product of nums[0..i-1]
    suffix = [1] * n   # suffix[i] = product of nums[i+1..n-1]
    result = [0] * n

    # Build prefix products
    for i in range(1, n):
        prefix[i] = prefix[i - 1] * nums[i - 1]

    # Build suffix products
    for i in range(n - 2, -1, -1):                         # note here we're using end index = -1 in range, to capture the 0th element i.e. nums[i + 1]
        suffix[i] = suffix[i + 1] * nums[i + 1]

    # Combine
    for i in range(n):
        result[i] = prefix[i] * suffix[i]

    return result

nums = [1,2,3,4]
print(product_except_self_prefix_suffix(nums))
````

**Time Complexity**: `O(n)`

**Space Complexity**: `O(n)` for prefix + `O(n)` for suffix + `O(n)` for result = `O(n)`
